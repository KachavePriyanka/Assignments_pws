```python
#5. Generate a synthetic dataset with correlated features. Visualize the correlation matrix of a dataset using a
#heatmap.
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


np.random.seed(0)
n_samples = 100
feature1 = np.random.rand(n_samples)
feature2 = 0.7 * feature1 + 0.1 * np.random.randn(n_samples) # Introduce some noise
feature3 = -0.5 * feature1 + 0.2 * np.random.randn(n_samples)
data = {'Feature1': feature1, 'Feature2': feature2, 'Feature3': feature3}
df = pd.DataFrame(data)


corr_matrix = df.corr()

plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix Heatmap')
plt.show()
```

```python
#2. Using the Student Grades, create a violin plot to display the distribution of scores across different grade
#categories.
import plotly.express as px
import plotly.graph_objects as go
import numpy as np
import pandas as pd
np.random.seed(15)
data = {
    'Grade': np.random.choice(['A', 'B', 'C', 'D', 'F'], 200),
    'Score': np.random.randint(50, 100, 200)
}
df = pd.DataFrame(data)

fig=px.violin(df,x='Grade',y='Score',color='Grade',title='Distribution of Scores Across Grade Categories')

fig.show()
```

```python
#3 Using the sales data, generate a heatmap to visualize the variation in sales across different months and days.

np.random.seed(20)
data = {
    'Month': np.random.choice(['Jan', 'Feb', 'Mar', 'Apr', 'May'], 100),
    'Day': np.random.choice(range(1, 31), 100),
    'Sales': np.random.randint(1000, 5000, 100)
}
df = pd.DataFrame(data)
fig=px.density_heatmap(df,x='Month',y='Day',z='Sales',title='Variation in Sales Across Months and Days')
fig.show()
```

```python
#4. Using the given x and y data, generate a 3D surface plot to visualize the function

x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
x, y = np.meshgrid(x, y)
z = np.sin(np.sqrt(x**2 + y**2))
data = {
    'X': x.flatten(),
    'Y': y.flatten(),
    'Z': z.flatten()
}
df = pd.DataFrame(data)
fig=px.scatter_3d(df,x='X',y='Y',z='Z',title='3D Surface Plot')
fig.show()
#surface plot using go
surface = go.Surface(z=z, x=x, y=y, colorscale='rainbow')
fig = go.Figure()
fig.add_trace(surface)
fig.show()
```
