```python
#Demonstrate three different methods for creating identical 2D arrays in NumPy.
#Provide the code for each method and the final output after each method

import numpy as np
arr1 = np.linspace(1, 10, 100).reshape(10,10)
print("Method 1: Using linspace and reshape")
print(arr1)

# Method 2: Using np.full
arr2 = np.full((10,10), np.linspace(1,10,100).reshape(10,10))
print("\nMethod 2: Using full")
print(arr2)


# Method 3: Using a list comprehension and np.array
arr3 = np.array([[i for i in np.linspace(1,10,10)] for _ in range(10)])
print("\nMethod 3: Using list comprehension")
arr3
```
