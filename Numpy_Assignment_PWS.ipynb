1. Explain the purpose and advantages of NumPy in scientific computing and data analysis. How does it
enhance Python's capabilities for numerical operations?

NumPy (Numerical Python) is a core library in Python used for numerical and scientific computing. It provides support for arrays, matrices, and various mathematical functions, making complex data manipulations easy and efficient.

Advantages of NumPy:

Array-based operations: It supports multi-dimensional arrays (ndarrays) that enable large data structures to be managed efficiently.
Memory efficiency: NumPy arrays consume less memory than Python lists, which improves performance, especially for large datasets.
Speed: Operations on NumPy arrays are much faster than similar operations on lists due to optimized algorithms and lower-level implementations.
Vectorized operations: Mathematical and logical operations can be performed on entire arrays without the need for explicit loops.
Interoperability: It integrates seamlessly with other libraries like Pandas, SciPy, and Matplotlib, enhancing data analysis capabilities.



2. Compare and contrast np.mean() and np.average() functions in NumPy. When would you use one over the
other?


np.mean(): Calculates the arithmetic mean (average) of elements across an array or specified axis. It does not allow weighted calculations.
np.average(): Also calculates the mean but has an option to specify weights for elements, enabling a weighted average calculation.
Use np.mean() for a simple average of values, and np.average() when you need to apply weights to different values.


3. Describe the methods for reversing a NumPy array along different axes. Provide examples for 1D and 2D
arrays.

1D Array: Use slicing ([::-1]) to reverse a 1D array.


arr_1d = np.array([1, 2, 3, 4, 5])
print(arr_1d[::-1])   

2D Array: Use np.flip() with an axis parameter to reverse along specific axes.

Horizontal Flip (reverse rows): np.flip(arr, axis=1)
Vertical Flip (reverse columns): np.flip(arr, axis=0)

arr_2d = np.array([[1, 2], [3, 4], [5, 6]])
print(np.flip(arr_2d, axis=1))  
print(np.flip(arr_2d, axis=0))

5. Define ndarrays in NumPy and explain their key features. How do they differ from standard Python lists?

An ndarray (n-dimensional array) is the core data structure in NumPy for holding homogeneous data.

Key Features:
Fixed size: Once created, the size of an ndarray cannot be changed.
Multidimensional: Supports multi-dimensional arrays, allowing for complex data structures.
Efficient operations: ndarrays allow element-wise and vectorized operations, improving computation speed.
Differentiation from lists: Unlike Python lists, ndarrays store elements in continuous memory blocks, making them faster for numerical operations.


6. Analyze the performance benefits of NumPy arrays over Python lists for large-scale numerical operations.

NumPy arrays are optimized for numerical computations because they:
Have contiguous memory allocation: Reduces the overhead seen in Python lists, which are pointers to separate objects.
Support vectorized operations: Minimize the need for loops in Python, thus reducing execution time.
Allow better cache utilization: CPU cache is used more efficiently, speeding up repeated accesses.


7. Compare vstack() and hstack() functions in NumPy. Provide examples demonstrating their usage and
output.

vstack(): Stacks arrays vertically (row-wise).
hstack(): Stacks arrays horizontally (column-wise).
Example:
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])

.....Vertical stack
print(np.vstack((arr1, arr2)))
....Horizontal stack
print(np.hstack((arr1, arr2)))


8. Explain the differences between fliplr() and flipud() methods in NumPy, including their effects on various
array dimensions.

fliplr(): Flips an array along the left-right axis (columns). Used for 2D arrays or higher.
flipud(): Flips an array along the up-down axis (rows). Also for 2D or higher dimensions.
Example:
arr = np.array([[1, 2], [3, 4], [5, 6]])
....Left-right flip
print(np.fliplr(arr))
....Up-down flip
print(np.flipud(arr))


9. Discuss the functionality of the array_split() method in NumPy. How does it handle uneven splits?

array_split() divides an array into specified subarrays, even if the split is uneven. It ensures that all sections are created, with some sections possibly having fewer elements.

Example:
arr = np.array([1, 2, 3, 4, 5])
print(np.array_split(arr, 3))

10. Explain the concepts of vectorization and broadcasting in NumPy. How do they contribute to efficient array
operations?

Vectorization: Allows operations on entire arrays without explicit loops, enhancing speed and code simplicity.
Broadcasting: Lets arrays with different shapes interact in element-wise operations by “stretching” smaller arrays to match the shape of the larger array.
Example:
arr = np.array([1, 2, 3])
print(arr + 5)

```python
# 10. Create a NumPy array representing daily temperatures for a month.
# Calculate and display the weekly averages
temp=np.random.randint(15,35,size=30)
temp
weekly_averages = []
for i in range(0, 30, 7):
  week_temps =temp[i:min(i + 7, 30)]
  weekly_average = np.mean(week_temps)
  weekly_averages.append(weekly_average)

print("Weekly Average Temperatures:")
for i, avg in enumerate(weekly_averages):
  print(f"Week {i+1}: {avg:.2f}")
```
